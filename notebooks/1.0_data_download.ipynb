# Download Data

## Purpose: Test the download of the 2 main datasets (Cap IV and adjunto)

- [Producción de Pozos de Gas y Petróleo No Convencional](https://datos.gob.ar/dataset/energia-produccion-petroleo-gas-por-pozo-capitulo-iv/archivo/energia_b5b58cdc-9e07-41f9-b392-fb9ec68b0725)
- [Datos de fractura de pozos de hidrocarburos (Adjunto IV)](https://datos.gob.ar/dataset/energia-datos-fractura-pozos-hidrocarburos-adjunto-iv)

```python
import pandas as pd
import requests
```

```python
url_production = "http://datos.energia.gob.ar/dataset/c846e79c-026c-4040-897f-1ad3543b407c/resource/b5b58cdc-9e07-41f9-b392-fb9ec68b0725/download/produccin-de-pozos-de-gas-y-petrleo-no-convencional.csv"
url_completion = "http://datos.energia.gob.ar/dataset/71fa2e84-0316-4a1b-af68-7f35e41f58d7/resource/2280ad92-6ed3-403e-a095-50139863ab0d/download/datos-de-fractura-de-pozos-de-hidrocarburos-adjunto-iv-actualizacin-diaria.csv"
```

```python
with requests.get(url_production, stream=True) as r:
    r.raise_for_status()
    for i, line in enumerate(r.iter_lines()):
        print(line.decode("utf-8"))
        if i == 9:
            break
```

```python
with requests.get(url_completion, stream=True) as r:
    r.raise_for_status()
    for i, line in enumerate(r.iter_lines()):
        print(line.decode("utf-8"))
        if i == 9:
            break
```

```python
df_prod = pd.read_csv(
    filepath_or_buffer=url_production,
    sep=",",
    header=0,
    true_values=["t"],
    false_values=["f"],
    decimal=".",
    encoding="utf-8"
)
```

```python
print(df_prod.head(10))
```

```python
df_comp = pd.read_csv(
    filepath_or_buffer=url_completion,
    sep=",",
    header=0,
    true_values=["t"],
    false_values=["f"],
    decimal=".",
    encoding="utf-8"
)
```

```python
df_prod.to_parquet("../data/01_external/production.parquet")
df_comp.to_parquet("../data/01_external/completion.parquet")
```
