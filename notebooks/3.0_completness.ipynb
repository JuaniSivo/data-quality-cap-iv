## Purpose: Understand how to find the metrics for the completness

```python
from pathlib import Path

import pandas as pd
```

```python
p_ext_data = Path("../data/01_external")
```

```python
df_prod = pd.read_parquet(p_ext_data.joinpath("production.parquet"))
df_comp = pd.read_parquet(p_ext_data.joinpath("completion.parquet"))
```

```python
print("\nMissing values:\n", df_prod.isnull().sum())
```

```python
col_complete_percentage = 1 - df_prod.isnull().sum() / df_prod.shape[0]
col_complete_percentage = col_complete_percentage * 100
print(col_complete_percentage.map("{:.2f}".format))
```
